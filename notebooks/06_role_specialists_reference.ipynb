# Fase 6: especialistas por rol para comunicación

## Quién se muestra en el radar chart, y cómo se etiqueta cada nombre

A diferencia de los notebooks 0-5, este no produce evidencia para la
pregunta central, produce un artefacto de comunicación: el pequeño
conjunto de jugadores que se muestra en el radar chart (ver el
planteamiento del proyecto, sección 13), que ilustra la diferencia entre
perfiles especializados y multidimensionales. Dos categorías, que nunca
se mezclan sin etiqueta: el líder estadístico de cada rol dentro de la
población calificada (dato), y jugadores de referencia agregados
manualmente y mostrados por su reconocibilidad aunque no lideren un rol
en esta población específica (decisión de comunicación), según lo ya
decidido en `METHODOLOGY.md` bajo "Líderes por rol / jugadores de
referencia".

```python
import pandas as pd

role_scores = pd.read_csv("../data/processed/role_scores.csv")

role_cols = ["finisher_score", "dribbler_score", "chance_creator_score", "play_organizer_score"]

print("Líder estadístico por rol (población calificada, n=2488):\n")
for role in role_cols:
    leader = role_scores.loc[role_scores[role].idxmax()]
    print(f"{role}: {leader['name']} ({leader[role]:.2f})")
    perfil = leader[role_cols].astype(float).round(2).to_dict()
    print(f"  Perfil completo: {perfil}")
    print()
```

```python
role_metrics_df = pd.read_csv("../data/processed/role_metrics.csv")

leaders_info = []
for role in role_cols:
    leader = role_scores.loc[role_scores[role].idxmax()]
    minutes = role_metrics_df.loc[role_metrics_df["id"] == leader["id"], "minutesPlayed"].values[0]
    world_cups = role_metrics_df.loc[role_metrics_df["id"] == leader["id"], "world_cups_played"].values[0]
    leaders_info.append({
        "role": role, "name": leader["name"], "score": leader[role],
        "minutesPlayed": minutes, "world_cups_played": world_cups,
    })

leaders_df = pd.DataFrame(leaders_info)
print(leaders_df.to_string(index=False))
```

## A stricter floor for choosing who gets called a "leader"

The organizer-role leader (Isco, 390 minutes) illustrates a real problem:
`totalOppositionHalfPasses` measures passing volume in the opposition
half, not whether that possession created danger, and Spain's 2018
round-of-16 exit to Russia is the textbook case of high-volume, low-threat
possession football. With only 390 minutes, a single match like that
dominates the per-90 rate.

The 270-minute population threshold isn't being changed, it's already
validated with real evidence in Phase 3 and reused throughout. What
changes here is narrower: for choosing which player gets featured as a
role's statistical leader in communication material, a stricter 900-minute
floor is applied (already established in Phase 5 as a meaningful
real-career-sample checkpoint, roughly 10 full matches). Players below
900 minutes stay in the analytical population and every finding already
reported, they're just not eligible to be showcased as "the leader" of a
role for the radar chart.

---

## Un piso más exigente para elegir a quién se llama "líder"

El líder de organizador (Isco, 390 minutos) ilustra un problema real:
`totalOppositionHalfPasses` mide volumen de pase en campo rival, no si esa
posesión generó peligro, y la eliminación de España en octavos del
Mundial 2018 contra Rusia es el caso de libro de fútbol de posesión de
alto volumen y bajo peligro. Con solo 390 minutos, un partido así domina
la tasa per 90.

El umbral poblacional de 270 minutos no se cambia, ya está validado con
evidencia real en la Fase 3 y reutilizado en todo el proyecto. Lo que
cambia acá es más acotado: para elegir qué jugador se muestra como líder
estadístico de un rol en el material de comunicación, se aplica un piso
más exigente de 900 minutos (ya establecido en la Fase 5 como un punto de
corte con sentido real de carrera, aproximadamente 10 partidos completos).
Los jugadores por debajo de 900 minutos siguen en la población analítica
y en todo hallazgo ya reportado, simplemente no son elegibles para
mostrarse como "el líder" de un rol en el radar chart.

```python
print("Columnas disponibles:", role_scores.columns.tolist())

qualified_900 = role_scores[role_scores["minutesPlayed"] >= 900]

print(f"\nJugadores elegibles como líder (900+ minutos): {len(qualified_900)} de {len(role_scores)}\n")

leaders_info = []
for role in role_cols:
    leader = qualified_900.loc[qualified_900[role].idxmax()]
    leaders_info.append({
        "role": role, "name": leader["name"], "score": leader[role],
        "minutesPlayed": leader["minutesPlayed"], "world_cups_played": leader["world_cups_played"],
    })

leaders_df = pd.DataFrame(leaders_info)
print(leaders_df.to_string(index=False))
```

```python
reference_pairs = {
    "play_organizer": ["Xavi", "Andrea Pirlo"],
    "chance_creator": ["Zinedine Zidane", "Johan Cruyff"],
    "dribbler": ["Neymar", "Jairzinho"],
    "finisher": ["Cristiano Ronaldo", "Kylian Mbappé"],
}

all_references = [name for pair in reference_pairs.values() for name in pair]

reference_df = role_scores[role_scores["name"].isin(all_references)][
    ["name", "minutesPlayed", "world_cups_played"] + role_cols
]
print(reference_df.to_string(index=False))

found = reference_df["name"].tolist()
missing = [n for n in all_references if n not in found]
print(f"\nNo encontrados en la población (revisar nombre exacto o si no llega a 270 min): {missing}")
```

```python
statistical_leaders = leaders_df.copy()
statistical_leaders["role_context"] = statistical_leaders["role"].str.replace("_score", "", regex=False)
statistical_leaders = statistical_leaders.drop(columns=["role"])
statistical_leaders["label"] = "statistical_leader"

reference_rows = []
for role, names in reference_pairs.items():
    for name in names:
        row = role_scores[role_scores["name"] == name].iloc[0]
        reference_rows.append({
            "role_context": role,
            "name": name,
            "score": row[f"{role}_score"],
            "minutesPlayed": row["minutesPlayed"],
            "world_cups_played": row["world_cups_played"],
        })

reference_leaders = pd.DataFrame(reference_rows)
reference_leaders["label"] = "reference_addition"

role_specialists = pd.concat([statistical_leaders, reference_leaders], ignore_index=True)
print(role_specialists.sort_values(["role_context", "label"], ascending=[True, False]).to_string(index=False))
```

```python
role_specialists.to_csv("../data/processed/role_specialists.csv", index=False)
print(f"Guardado: {role_specialists.shape[0]} filas, {role_specialists.shape[1]} columnas")
```

## Top 10 por rol, de referencia

No se usa en el resto del pipeline, se deja acá simplemente como registro
visual de cómo se ve el top de la distribución de cada rol, tanto en el
umbral analítico de 270 minutos como en el piso más exigente de 900
minutos usado para elegir a los líderes estadísticos de este notebook.

```python
qualified_900 = role_scores[role_scores["minutesPlayed"] >= 900]

for role in role_cols:
    print(f"\n=== Top 10: {role} (270+ min) ===")
    top10_270 = role_scores.nlargest(10, role)[["name", role, "minutesPlayed", "world_cups_played"]]
    print(top10_270.to_string(index=False))

    print(f"\n=== Top 10: {role} (900+ min) ===")
    top10_900 = qualified_900.nlargest(10, role)[["name", role, "minutesPlayed", "world_cups_played"]]
    print(top10_900.to_string(index=False))
```

## Phase 6 conclusion

Built the reference set for the radar chart: one statistical leader per
role (David Villa, Maradona, Beckham, Rodri, all with a 900-minute floor
applied specifically for this selection, stricter than the 270-minute
analytical population, after Isco's 390-minute organizer lead turned out
to be driven by Spain's low-threat, high-volume passing display in the
2018 round of 16 against Russia) plus two manually added reference

**Result:** `data/processed/role_specialists.csv`, 12 rows, explicitly
labeled `statistical_leader` or `reference_addition`, ready to feed the
Tableau radar chart without ambiguity between data and communication
choices.
